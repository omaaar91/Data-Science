```python

df['price'] = df['price'].str.replace('£', '').str.strip().astype(float)
```

```python
df['price'].dtype
```

```python
df.info()
```

```python
rating_dictionary = {
    'one' : 1,
    'two' : 2,
    'three':3,
    'four' : 4,
    'five': 5
}
```

```python
df['rating'] = df['rating'].astype(str).str.strip().str.lower().map(rating_dictionary)
```

```python
df['rating'].dtype
```

```python
df.head()
```

```python
df.info()
```

```python
df['availability'] = df['availability'].str.extract(r'(\d+)').astype(int)
```

```python
import pandas as pd
from sqlalchemy import create_engine
import urllib

# 1. الاتصال بقاعدة البيانات
conn_str = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=localhost;"
    "DATABASE=Ecommerce;"
    "Trusted_Connection=yes;"
)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={urllib.parse.quote_plus(conn_str)}")

# 2. قراءة كل البيانات المدمجة وتصديرها إلى CSV
df_master = pd.read_sql("SELECT * FROM v_Ecommerce_Master", con=engine)
df_master.to_csv("Ecommerce_Master.csv", index=False)
print(f"✅ تم تصدير {len(df_master)} سجل في ملف 'Ecommerce_Master.csv' بنجاح!")
```
